#### 1.2.3. Pandas读取HTML数据方法

pd.read_html()方法
- 只能读取网页中的表格(即<table>标签内的数据)    

pd.read_html(io, *[, match, flavor, header, ...])  - options

参数说明：  
- io: 字符串（用单/双引号），被导入的带路径的文件名，也可以是URL链接.注意：网址不接受https
- match: 字符串，正则表达式
- flavor：解释器。默认“lxml”
- header：整数，指定列名所在的行数。默认值：0. 无列名：None
- encoding：字符串，指定文件的编码格式。


```python
import urllib.request

url = 'http://en.wikipedia.org/wiki/2008_Summer_Olympics'
headers = {'User-Agent': 'Mozilla/5.0'}
req = urllib.request.Request(url, headers=headers)   #return a request object

# turn request object into file like object or string, in oder to read it with pd.read_html
with urllib.request.urlopen(req) as response:
    html = response.read()

medal_table = pd.read_html(html, match='2008 Summer Olympics medal table')

print(medal_table) 
```

#### 1.2.6. Pandas读取Excel Table数据方法

pd.read_excel() - Options

>**pandas.`read_excel`**(<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;io,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;sheet_name=0,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;header=0,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;usecols=None,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;nrows=0,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;true_values=None,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;skiprows=None，<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;skipfooter=0，<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;converters=None,<br> 
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;dtype=None,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;)

参数说明：
- io： 直接使用pd.read_excel(r"文件路径")，默认读取第一个sheet的全部数据。 支持str, bytes, ExcelFile, xlrd.Book, path object, or file-like object
- sheet_name：可选（str, int, list, None, default 0）
  - str: 字符串用于引用的sheet的名称(如果有特定名称)
  - int: 整数用于引用的sheet的隐性索引（从0开始）
  - list：整数和字符串组成的列表，读取后的数据类型是OrderedDict,将列表指定的sheet的数据合并到了一个list中
  - None：表示引用所有sheet （相当于=list（all）的结果）
  - 0： 默认第一个sheet
- header: 表示用第几行作为表头，可选（int, list of int, default 0）
  - 0： 默认，即第一行为表头 
  - int: 跳选第int+1行为表头，前面行数据就被忽略了。行数从0开始
  - [int]：选择列表int指定行的数据作为表头。可用于读取合并行数据做表头
  - None: 不用源表头
- usecols:可选(int, str, list-like, or callable default None)
  - None：默认值，解析所有列。
  - str： 表示Excel列字母。列范围的逗号分隔列表（例如“ A：E”或“ A，C，E：F”）。`范围全闭`。
  - int: 解析到第几列。`对应隐性列索引，从0开始，但是范围全闭（容易让人抓狂的地方）`。 如usecols=3: 选0，1，2，3共4列数据
  - int列表: 解析int对应的那几列。
- index_col：指定列为索引列，可选（int, list of int, default None）
  - 默认为None，即隐性的行作DataFrame的行标签。
  - int: 指定第几列为索引列. 从0开始。
  - list of int：选择列表中的整数列为索引列（多重列索引）
- nrows：（int, default None），指定需要读取前多少行，通常用于较大的数据文件中。注意：从header行下一行开始算！
  - 默认为None
  - int：nrows=3 读取前三行
- skiprows：跳过指定的行，可选（int，list like）
  - int: 跳过前int行=从int+1行开始读取
  - [int]：skiprows=[1,3,5] 跳过第1,3,5行
- dtype: 指定读入的列的数据类型。可选（Type name or dict of column -> type, default None）
  - 默认为None，也就是不改变数据类型
  - 指定列的类型名称或字典。
  - 场景：作用强大，强制限制数据类型，避免后续处理出现意外。如原始字符串数字被当成数值进行运算。可结合df.info先查看各列数据类型是否合乎要求

```python

# Attetion: String cannot perform math operations!
test_op = '20%' * 2
print(f"no math operation for string %: {test_op}")

# Create Function to covert % to float
def convert_string_to_float(row):
    '''Convert string (%) to float'''
    pct_value = row.replace('%','')
    pct_value = float(pct_value) / 100
    return pct_value

# Converters: Dict of functions for converting values in certain columns
df = pd.read_excel("data/complex_tables.xlsx",
                   sheet_name='Orders',
                   header=5,
                   usecols='B:C,E:N',
                   true_values=['Yes'],
                   skipfooter = 2,
                   converters={'Discount': convert_string_to_float}
                   )
df.head()
```

### 1.2.7. Pandas读取XML数据 

XML（eXtensible Markup Language）是一种标记语言，广泛应用于数据存储、数据交换和配置文件等领域。特征：1. 每个数据都用Tags框住，文本和HTML格式相似。但没有预设的Tags， 2. 具有特定结构：可以多级内嵌, 3. 基本组成：元素（Element）和属性（Attribute）：可以理解为是要被读取的数据和相应的说明。

XML数据通常的处理方法：
1. 组合法：先用Python内置或第三方库来解析XML文本，找出元素和属性，再将找到的元素和属性交给pd.DataFrame()转换成df
2. 直接使用Pandas的read_xml()

玩转Pandas的Index标签  
1. 知识点和理解框架：
- 对Index标签本身的操作：
  - 你想做什么：
    - 构建
    - 修改
    - 删除
    - 指定/设置标签，  重制标签
  - 你得知道怎么做：
    - df.set_index(), df.reset_index()
- 用标签（Index）来对数据的操作=索引（Indexing）：
  - 你想做什么：
    - 筛选（整串， 或局部=切片）
    - 排序
    - 添加/删除
    - 转换
    - 分组/整合/拼接
  - 你得知道怎么做：
    - df[], df.loc(), df.iloc()


1. 运用：
   - 从简单到复杂地练习
   - 对大家有帮助的一个方法: 
     - 选择df[], df.loc(), df.iloc()的逻辑分析框架：
    问：自己是否知道数据的固定范围？ 答是的，“出门向左”，继续走1。答否的，“出门向右”，问：自己是否知道想要怎样的数据， 但不知道在表格里都是分布/位置？ 答是的，“一路向北”，继续2。答否的：“更上一层楼”，思考是否任务太大，需要分解目标。
     1. 自己知道数据的固定位置或范围
     2. 自己不确定数据位置/范围，需要采用某种筛选条件来选择
     3. 自己不知道如何来框定数据，想法太复杂（“我想要一份红烧排骨来炖干煸排骨汤”），思考你的思考

##### 1.3.2.1. 索引生命周期管理（Index本身的操作）

| 操作类型 | 目的 | 核心方法 | 代码示例 |
|----------|------|----------|----------|
| 创建构建 | 从数据创建索引   | `pd.index()`, `set_index()` | `df.set_index('日期')` |
| 修改调整 | 重命名/替换索引  | `rename()`, `set_index()` | `df.rename(index={'old':'new'})` |
| 重置恢复 | 索引变列，恢复默认 | `reset_index()` | `df.reset_index(drop=True)` |
| 删除清理 | 移除索引        | `reset_index(drop=True)` | `df.reset_index(drop=True)` |
| 类型转换 | 改变索引数据类型 | `astype()`, `pd.to_datetime()` | `df.index.astype('category')` |

- set_index() 将已存在的列标签(对应内容)设置为 DataFrame 行索引。除了可以添加索引外，也可以替换已经存在的索引。常见场景：把 Series 或者一个 DataFrame 设置成另一个 DataFrame 的索引

```python
# 在 iloc 索引器中结合使用掩码与的索引方法  
data.iloc[data.density > 100, [1, 2]]  # iloc 只能使用索引位置
```

```python
# Option 1: Use iloc with integer positions
mask = data.density > 100
data.iloc[mask.values, [1, 2]]

# Option 2: Use loc instead (more readable)
data.loc[data.density > 100, data.columns[[1, 2]]]

# Option 3: Chain the operations
data[data.density > 100].iloc[:, [1, 2]]
```

### 以列为对象，对数据表中的数据进行数学运算。 

```python
# 1. 新增一列：总人口数增加1000
df['pop_plus_1000'] = df['pop'] + 1000

# 2. 新增一列：面积减少10%
df['area_minus_10pct'] = df['area'] * 0.9

# 3. 新增一列：人口密度翻倍
df['density_double'] = df['density'] * 2

# 4. 新增一列：人口除以面积的整数部分
df['pop_div_area_int'] = df['pop'] // df['area']

# 5. 新增一列：人口对面积取余
df['pop_mod_area'] = df['pop'] % df['area']

# 6. 新增一列：人口密度的平方
df['density_pow2'] = df['density'] ** 2

# 7. 新增一列：人口密度大于100的布尔值
df['density_gt_100'] = df['density'] > 100

# 8. 新增一列：人口密度小于50或大于200
df['density_outlier'] = (df['density'] < 50) | (df['density'] > 200)

# 9. 新增一列：人口密度等于最大密度
df['density_is_max'] = df['density'] == df['density'].max()

# 10. 新增一列：人口密度排名（降序）
df['density_rank'] = df['density'].rank(ascending=False, method='min').astype(int)

display(df.head())
```
