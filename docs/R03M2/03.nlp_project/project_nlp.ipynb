NLP文本预处理方法小总结:
将原始文本转换为机器可理解的格式，是NLP的基石：
- 分词 (Tokenization): 将句子拆分为单词或词组。
- 去停用词 (Stop Words Removal): 移除“的”、“is”、“a”等高频但无实际意义的词。
- 词形还原 (Lemmatization) / 词干提取 (Stemming): 将单词还原为根形式。
- 词性标注 (POS Tagging): 识别词语的词性（名词、动词、形容词）

经典处理方法：re


```python
import re

#Find all will return all matching strings
#In this case we will get back just the exact string because it is what we are looking for
print(re.findall("rä", test_string))
```

```python
#By using [a-z], we signal the next piece can be any lower case letter a-z
print(re.findall("at[a-z]", test_string))
```

```python
#Or using limitation to narrow the search, for example [a-r] we limit to letters a through r
print(re.findall("PS-[B-W][B-W]", test_string))
```

```python
#For case sensitive char searching 
print(re.findall("aa[a-zA-Z]", test_string))
```

```python
#The split function splits text by the searched text
print(re.split("Chin[a-zA-Z]", test_string))
```

```python
#Searching for numbers: Notice the 33 will be split
print(re.findall("[0-9]",test_string))
```

### 3.2. Text Data Analysis using NLP libs



```python
import nltk
#nltk is the library we will use for text analysis
stemmer = nltk.stem.SnowballStemmer('english')  # german is also supported 
```

```python
#The stemmer breaks down words into roots
stemmer.stem("run running")
```

```python
words = "run running runs runner ran".split()
stems = [stemmer.stem(word) for word in words]
print(stems)
```

Remark: 
- The SnowballStemmer in NLTK converts all input to lowercase before stemming, regardless of the original case. This is standard for most stemmers, including German, to ensure consistent stemming results.

- German nouns are always capitalized, but the stemmer ignores case to treat words like Haus and haus the same.

```python
# stem the test_string
stemmer = nltk.stem.SnowballStemmer('german')
words = test_string.split()
stems = [stemmer.stem(word) for word in words]
print(stems)
```

**sklearn function `CountVectorizer`**

CountVectorizer is a feature extraction tool from scikit-learn that converts a collection of text documents to a matrix of token counts (bag-of-words model). Each row represents a document, and each column represents a unique word (token) from the corpus. The cell values are the counts of each word in each document.

```python
#To get counts, we can use the count vectorizer
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()

# test_text = "Text that we want to analyze with text analysis."
test_text = stemmer.stem(test_string)
print(test_text)
print()
#The text is not readable from the initial fit_transform because it is in sparse matrix form
counts = vectorizer.fit_transform([test_text])
print(counts)
```

```python
#To array gives us the counts but not the words
print(counts.toarray())
```

```python
#The function get_feature_names_out returns the names
print(vectorizer.get_feature_names_out())
```

```python
#Convert the array to a series
print(pd.Series(counts.toarray()[0], index = vectorizer.get_feature_names_out()))
```

```python
#Let's turn these text treatment steps into a function and add more

def get_text(text:str):
    stemmer = nltk.stem.SnowballStemmer('german')
    #stems = [stemmer.stem(word) for word in words]
    stems = stemmer.stem(text)
    # Example German stop words (expand as needed)
    german_stop_words = ["und", "oder", "aber", "der", "die", "das", "ein", "eine", "es","den", "dem", "des",
                         "ist", "war", "im", "in", "zu", "mit", "auf", "fur", "von","so", "werden", "wurde"]
    vectorizer = CountVectorizer(stop_words=german_stop_words)
    counts = vectorizer.fit_transform([stems])
    counts = pd.Series(counts.toarray()[0],index=vectorizer.get_feature_names_out())
    return counts

counts = get_text(test_string)
print(counts.sort_values(ascending=False))
```

```python
#Find the most common words
print(counts.sort_values(ascending=False).head(100))
```

```python
#Convert to frequency
word_frequency = counts.copy()
word_frequency = word_frequency / word_frequency.sum()
print(word_frequency)
```

```python
# find the distance between two words
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

word1 = "produktion"
word2 = "wir"

# Create vectors for each word (as 1D arrays)
vec1 = np.array([[counts[word1]]])
vec2 = np.array([[counts[word2]]])

# Cosine similarity
similarity = cosine_similarity(vec1, vec2)[0][0]
print("Cosine similarity:", similarity)

# Euclidean distance
distance = np.linalg.norm(vec1 - vec2)
print("Euclidean distance:", distance)
```

```python
print("\n--- Starting German Tokenization ---")
print(f"Original Text:\n{test_string}\n")

# 1. Sentence Tokenization for German
sentences = nltk.sent_tokenize(test_string, language='german')
print("--- Sentence Tokens ---")
for i, sentence in enumerate(sentences):
    print(f"Satz {i+1}: {sentence}") # "Satz" is German for "Sentence"

# 2. Word Tokenization for German
words = nltk.word_tokenize(test_string, language='german')
print("\n--- Word Tokens ---")
print(words)
```

```python
# Stemming
stemmer = nltk.stem.PorterStemmer()
stems = [stemmer.stem(word) for word in words]
print('Stems:', stems)

# Lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in words]
print('Lemmas:', lemmas)

# POS Tagging
pos_tags = nltk.pos_tag(words)
print('POS tags:', pos_tags)
```

The POS Tagging output flags are short codes that represent the part of speech (POS) assigned to each word by NLTK's tagger. For example, in the output:

Each tuple is (word, tag). Here are some common POS tags and their meanings:
- NN: Noun, singular (e.g., "dog")
- NNS: Noun, plural (e.g., "dogs")
- NNP: Proper noun, singular (e.g., "Python")
- VB: Verb, base form (e.g., "run")
- VBD: Verb, past tense (e.g., "ran")
- VBG: Verb, gerund/present participle (e.g., "running")
- VBN: Verb, past participle (e.g., "eaten")
- VBP: Verb, non-3rd person singular present (e.g., "run")
- VBZ: Verb, 3rd person singular present (e.g., "runs")
- JJ: Adjective (e.g., "big")
- JJR: Adjective, comparative (e.g., "bigger")
- JJS: Adjective, superlative (e.g., "biggest")
- RB: Adverb (e.g., "quickly")
- IN: Preposition/subordinating conjunction (e.g., "in", "for")
- DT: Determiner (e.g., "the", "a")
- PRP: Personal pronoun (e.g., "he", "she")
- PRP$: Possessive pronoun (e.g., "his", "her")
- CC: Coordinating conjunction (e.g., "and", "but")

For a full list, see the Penn Treebank POS tag set:

```python
# List of German sentences to analyze
german_sentences = [
    "Das ist ein wunderbares und brillantes Produkt!",    # Clearly positive
    "Ich habe heute einen wirklich schrecklichen Tag.",    # Clearly negative
    "Das Wetter ist heute bewölkt.",                       # Neutral
    "Der Film war eigentlich gar nicht so schlecht.",       # A more complex case
    "Diese Lösung ist absolut fantastisch und sehr hilfreich." # Another positive
]

# List of German sentences to analyze
def analyze_german_sentiment_flair(text):
    from flair.data import Sentence
    from flair.models import TextClassifier

    classifier = TextClassifier.load('sentiment')
    print(f"\nAnalyzing: '{text}'")
    sentence = Sentence(text)  # text must be a string
    classifier.predict(sentence)
    label = sentence.labels[0]
    print(f"Verdict: {label.value} (Confidence: {label.score:.2f})")



# Run the Flair analysis
analyze_german_sentiment_flair(german_sentences)
```

### Classic NLP Treatments (No LLM)

This section demonstrates basic NLP tasks using traditional Python libraries:

- **Tokenization**: Splitting text into words/sentences
- **Stemming**: Reducing words to their root form
- **Lemmatization**: Reducing words to their base form (requires POS tagging)
- **POS Tagging**: Part-of-speech tagging
- **Sentiment Analysis**: Using TextBlob or Vader (English only)

These methods work without large language models and are suitable for simple text analysis.

```python
# Sentiment Analysis (English only, classic method)
from textblob import TextBlob

blob = TextBlob(test_string)
print('Polarity:', blob.sentiment.polarity)
print('Subjectivity:', blob.sentiment.subjectivity)

# Vader sentiment (English only, more robust for social media)
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')
sia = SentimentIntensityAnalyzer()
print('Vader sentiment:', sia.polarity_scores(test_string))
```
